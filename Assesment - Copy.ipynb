```python

kkr_matches = df[(df['team1'] == 'Kolkata Knight Riders') | (df['team2'] == 'Kolkata Knight Riders')]


umpire1 = kkr_matches['umpire1'].value_counts()
umpire2= kkr_matches['umpire2'].value_counts()


total_umpire_counts = (umpire1 + umpire2).fillna(0).astype(int)


most_frequent_umpire = total_umpire_counts.idxmax()

print(most_frequent_umpire)
```
